```python
df = pd.read_csv("cohort_index_dataset.csv")
```

```python
cohort_data = df.groupby(
    ["Cohort_Month", "Cohort_Index"]
)["Customer ID"].nunique().reset_index()
```

```python
cohort_matrix = cohort_data.pivot(
    index="Cohort_Month",
    columns="Cohort_Index",
    values="Customer ID"
)
```

```python
cohort_matrix.to_csv("retention_matrix.csv")
```

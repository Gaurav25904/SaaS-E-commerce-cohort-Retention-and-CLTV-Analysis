```python
df = pd.read_csv("cohort_index_dataset.csv")
```

```python
df.columns
```

```python
total_revenue = df["Revenue"].sum()

total_orders = df["Invoice"].nunique()

aov = total_revenue / total_orders

print("Average Order Value (AOV):", round(aov, 2))
```

```python
total_customers = df["Customer ID"].nunique()

purchase_frequency = total_orders / total_customers

print("Purchase Frequency:", round(purchase_frequency, 2))
```

```python
customer_lifespan = 12

cltv = aov * purchase_frequency * customer_lifespan

print("Customer Lifetime Value (CLTV):", round(cltv, 2))
```

```python
metrics = pd.DataFrame({
    "Metric": [
        "Total Revenue",
        "Total Orders",
        "Total Customers",
        "AOV",
        "Purchase Frequency",
        "CLTV"
    ],
    "Value": [
        total_revenue,
        total_orders,
        total_customers,
        aov,
        purchase_frequency,
        cltv
    ]
})

metrics
```

```python
metrics.to_csv(
    "customer_metrics.csv",
    index=False
)
```

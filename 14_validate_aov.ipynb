```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("cohort_index_dataset.csv")
```

```python
total_revenue = df["Revenue"].sum()

total_orders = df["Invoice"].nunique()

aov = total_revenue / total_orders

print("Total Revenue:", round(total_revenue, 2))
print("Total Orders:", total_orders)
print("Average Order Value (AOV):", round(aov, 2))
```

```python
negative_revenue = df[df["Revenue"] < 0]

print("Negative Revenue Rows:", len(negative_revenue))
```

```python
duplicate_invoices = df["Invoice"].duplicated().sum()

print("Duplicate Invoices:", duplicate_invoices)
```

```python
total_customers = df["Customer ID"].nunique()

print("Total Customers:", total_customers)
```

```python
validation_summary = pd.DataFrame({
    "Metric": [
        "Total Revenue",
        "Total Orders",
        "Total Customers",
        "AOV",
        "Negative Revenue Rows",
        "Duplicate Invoices"
    ],
    "Value": [
        total_revenue,
        total_orders,
        total_customers,
        aov,
        len(negative_revenue),
        duplicate_invoices
    ]
})

validation_summary
```

```python
validation_summary.to_csv(
    "aov_validation_report.csv",
    index=False
)
```

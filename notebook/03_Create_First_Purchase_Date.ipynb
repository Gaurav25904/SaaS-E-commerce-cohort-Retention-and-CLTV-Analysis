```python
import pandas as pd

df = pd.read_csv("cleaned_online_retail.csv")
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df["FirstPurchaseDate"] = (
    df.groupby("Customer ID")["InvoiceDate"]
      .transform("min")
)
```

```python
import pandas as pd
```

```python
df = pd.read_csv("cohort_index_dataset.csv")
```

```python
df.columns
```

```python
df[["Customer ID","Cohort_Month","Order_Month","Cohort_Index"]].head(10)
```

```python
cohort_data = (
    df.groupby(["Cohort_Month", "Cohort_Index"])["Customer ID"]
      .nunique()
      .reset_index()
)
```

```python
cohort_data.head(10)
```

```python
cohort_matrix = cohort_data.pivot(
    index="Cohort_Month",
    columns="Cohort_Index",
    values="Customer ID"
)
```

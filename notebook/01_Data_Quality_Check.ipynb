# 01_Data_Quality_Check

This notebook checks the dataset structure, missing values, and duplicate records.

## Import Pandas

```python
import pandas as pd
```

## Load Excel File

```python
excel_file = pd.ExcelFile('online_retail_II.xlsx')
print(excel_file.sheet_names)
```

## Read Required Sheet

```python
# Replace with the sheet you want to analyze
df = pd.read_excel('online_retail_II.xlsx', sheet_name=excel_file.sheet_names[0])
```

## Preview Data

```python
df.head()
```

## Dataset Shape

```python
print('Shape:', df.shape)
```

## Dataset Information

```python
df.info()
```

## Missing Values

```python
missing = df.isnull().sum()
print(missing)
```

## Missing Value Percentage

```python
missing_pct = (df.isnull().mean()*100).round(2)
print(missing_pct)
```

## Columns with Missing Values

```python
print(missing[missing>0])
```

## Duplicate Rows

```python
print('Duplicate rows:', df.duplicated().sum())
```

## View Duplicate Records

```python
df[df.duplicated()].head()
```

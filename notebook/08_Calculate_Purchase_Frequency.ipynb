```python
import pandas as pd
```

```python
df = pd.read_csv("cohort_index_dataset.csv")
```

```python
df[["Customer ID", "Invoice"]].head()
```

```python
customer_orders = (
    df.groupby("Customer ID")["Invoice"]
      .nunique()
      .reset_index()
)
```

```python
customer_orders.columns = ["Customer ID", "Total_Purchases"]
```

```python
customer_orders.head()
```

```python
purchase_frequency = customer_orders["Total_Purchases"].mean()
```

```python
print("Purchase Frequency:", round(purchase_frequency, 2))
```

```python
import pandas as pd
```

```python
df = pd.read_csv("cohort_index_dataset.csv")
```

```python
df[["Customer ID", "InvoiceDate"]].head()
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
customer_lifespan = (
    df.groupby("Customer ID")["InvoiceDate"]
      .agg(["min", "max"])
      .reset_index()
)
```

```python
customer_lifespan.columns = [
    "Customer ID",
    "First_Purchase",
    "Last_Purchase"
]
```

```python
customer_lifespan["Lifespan_Days"] = (
    customer_lifespan["Last_Purchase"]
    - customer_lifespan["First_Purchase"]
).dt.days
```

```python
customer_lifespan["Lifespan_Months"] = (
    customer_lifespan["Lifespan_Days"] / 30
).round(2)
```

```python
customer_lifespan.head()
```

```python
average_lifespan = customer_lifespan["Lifespan_Months"].mean()

print("Average Customer Lifespan:", round(average_lifespan, 2), "months")
```

```python
customer_lifespan.to_csv("customer_lifespan.csv", index=False)
```

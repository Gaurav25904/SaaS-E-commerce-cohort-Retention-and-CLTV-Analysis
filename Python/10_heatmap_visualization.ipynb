```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
retention_rate = pd.read_csv(
    "retention_rate_dataset.csv",
    index_col="Cohort_Month"
)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    retention_rate,
    annot=True,
    fmt=".1f",
    cmap="Blues"
)

plt.title("Customer Retention Heatmap")
plt.xlabel("Cohort Index")
plt.ylabel("Cohort Month")

plt.show()
```

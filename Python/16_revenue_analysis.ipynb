```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cohort_index_dataset.csv")
```

```python
monthly_revenue = df.groupby(
    ["Year", "Month"]
)["Revenue"].sum().reset_index()

monthly_revenue
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_revenue.index,
    monthly_revenue["Revenue"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month Number")
plt.ylabel("Revenue")

plt.show()
```

```python
top_customers = df.groupby(
    "Customer ID"
)["Revenue"].sum().sort_values(
    ascending=False
).head(10)

top_customers
```

```python
plt.figure(figsize=(12, 6))

top_customers.plot(
    kind="bar"
)

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Customer ID")
plt.ylabel("Revenue")

plt.show()
```

```python
cohort_revenue = df.groupby(
    "Cohort_Month"
)["Revenue"].sum().reset_index()

cohort_revenue
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    cohort_revenue["Cohort_Month"],
    cohort_revenue["Revenue"],
    marker="o"
)

plt.xticks(rotation=90)

plt.title("Revenue by Cohort Month")
plt.xlabel("Cohort Month")
plt.ylabel("Revenue")

plt.show()
```

```python
monthly_revenue.to_csv(
    "monthly_revenue.csv",
    index=False
)

cohort_revenue.to_csv(
    "cohort_revenue.csv",
    index=False
)
```

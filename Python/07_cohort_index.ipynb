```python
df = pd.read_csv("New Cohort_dataset.csv")
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["First_Purchase_Date"] = pd.to_datetime(df["First_Purchase_Date"])
```

```python
invoice_year = df["InvoiceDate"].dt.year
invoice_month = df["InvoiceDate"].dt.month

cohort_year = df["First_Purchase_Date"].dt.year
cohort_month = df["First_Purchase_Date"].dt.month
```

```python
df["Cohort_Index"] = (
    (invoice_year - cohort_year) * 12
    + (invoice_month - cohort_month)
    + 1
)
```

```python
import pandas as pd

df = pd.read_csv("cohort_index_dataset.csv")
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
print("Minimum Revenue:", df["Revenue"].min())
print("Maximum Revenue:", df["Revenue"].max())
print("Average Revenue:", round(df["Revenue"].mean(),2))
print("Total Revenue:", round(df["Revenue"].sum(),2))
```

```python
print("Total Customers:", df["Customer ID"].nunique())
print("Total Orders:", df["Invoice"].nunique())
```

```python
summary = {
    "Rows": len(df),
    "Columns": len(df.columns),
    "Customers": df["Customer ID"].nunique(),
    "Orders": df["Invoice"].nunique(),
    "Revenue": round(df["Revenue"].sum(),2)
}

summary
```
